```python
from datasets import load_dataset
```

```python
ds = load_dataset("ingeniumacademy/reuters_articles")
```

```python
ds
```

```python
ds['train']
```

```python
ds['train'][0]
```

```python
def create_full_article_col(example):
    title_body = f"TITLE: {example['title']}\n\nBODY: {example['body']}"
    return {'full_article': title_body}
```

```python
ds_new = ds.map(create_full_article_col)
ds_new
```

```python
ds_new['train'][0]
```

```python
len_train_ds = len(ds_new['train'])
len_train_ds
```

```python
train_corpus = (ds_new['train'][i:i+1000]['full_article'] for i in range(0, len_train_ds, 1000))
train_corpus
```

```python
next(train_corpus)[4]
```

```python
from transformers import AutoTokenizer
```

```python
ckpt = 'gpt2'
ckpt
```

```python
tokenizer_1 = AutoTokenizer.from_pretrained(ckpt)
tokenizer_1
```

```python
tokenizer_2 = tokenizer_1.train_new_from_iterator(train_corpus, tokenizer_1.vocab_size)
#We do not want to change vocab size else it will create issues loading gpt2 model and we do not to change its
#token embedding input size
```

```python
tokenizer_2
```

```python
sample = ds_new['test'][2]['full_article']
sample
```

```python
tokenizer_1.tokenize(sample)
```

```python
tokenizer_2.tokenize(sample)
```

So the procedure used was as follows:

- Load a pretrained tokenizer
- Load dataset that you want to train it on and preprocess it as appropriate
- Create an Iterator for the dataset perhaps a generator object
- Create new tokenizer by feeding iterator and vocabulary size to train_new_from_iterator method
- Its preferable to keep vocab size same as the in original tokenizer unless you want to modify the input token embedding size in the original model architecture
- Save the trained tokenizer using save_pretrained method
